#### Latitude Longitude cleaning

```python
invalid_lat = ~dataset['latitude'].between(-90, 90)
invalid_lon = ~dataset['longitude'].between(-180, 180)

print(invalid_lat.sum())
print(invalid_lon.sum())
```

#### Time Zone cleaning

```python
dataset['timezone'] = dataset['timezone'].str.strip()
#print(dataset['timezone'].unique().tolist())
```

#### last updated epoch

```python
dataset['last_updated'] = pd.to_datetime(dataset['last_updated'], format='%d-%m-%Y %H:%M')
dataset['last_updated'] = dataset['last_updated'].dt.strftime('%d-%m-%Y %H:%M')
dataset.head(2)
```

#### temperature_celcius, temperature_fahrenheit, condition_text. 

```python
type(dataset['temperature_celsius'])
dataset = dataset.drop(columns=['temperature_fahrenheit'])
```

```python
dataset['temperature_celsius'] = dataset['temperature_celsius'].astype(float)
dataset['condition_text'] = dataset['condition_text'].astype('string').str.strip().str.title()
dataset.head(5)
```

#### Dropping redundant columns

```python
redundant_cols = [ 'feels_like_fahrenheit',
    'wind_mph', 'pressure_in', 'precip_in', 'visibility_miles', 'gust_mph'
]
dataset = dataset.drop(columns=[col for col in redundant_cols if col in dataset.columns])
print(dataset.dtypes)
```

#### changing data types

```python
cols_to_string = [
    'wind_direction',
    'sunrise', 'sunset', 'moonrise', 'moonset', 'moon_phase'
]

for col in cols_to_string:
    if dataset[col].dtype != 'string':
        dataset[col] = dataset[col].astype('string')

print(dataset.dtypes)
```

#### Monthly averages

```python
import pandas as pd
dataset['last_updated'] = pd.to_datetime(dataset['last_updated'], format='%d-%m-%Y %H:%M', errors='coerce')
dataset['year_month'] = dataset['last_updated'].dt.to_period('M')
num_cols = [
    'temperature_celsius', 'wind_kph', 'wind_degree', 'pressure_mb', 'precip_mm',
    'humidity', 'cloud', 'feels_like_celsius', 'visibility_km', 'uv_index',
    'gust_kph', 'air_quality_Carbon_Monoxide', 'air_quality_Ozone',
    'air_quality_Nitrogen_dioxide', 'air_quality_Sulphur_dioxide',
    'air_quality_PM2.5', 'air_quality_PM10'
]
monthly_avg = dataset.groupby(['country', 'location_name', 'year_month'])[num_cols].mean().reset_index()
monthly_avg.to_csv("weather_data_monthly.csv", index=False)

```

#### Yearly Averages

```python
import pandas as pd
dataset['last_updated'] = pd.to_datetime(dataset['last_updated'], errors='coerce')
dataset['year'] = dataset['last_updated'].dt.year
yearly_avg = dataset.groupby(['country', 'location_name', 'year']).mean(numeric_only=True).reset_index()
yearly_avg.to_csv("weather_yearly_avg.csv", index=False)
```

#### Seasonal Aggregation

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

dataset['month'] = dataset['last_updated'].dt.month
dataset['year'] = dataset['last_updated'].dt.year
dataset['season'] = dataset['month'].apply(get_season)
seasonal_avg = dataset.groupby(['country', 'location_name', 'year', 'season']).mean(numeric_only=True).reset_index()
seasonal_avg.to_csv("weather_seasonal_avg.csv", index=False)

```
